```python
import pandas as pd
```

```python
evidences = pd.read_json("/content/drive/MyDrive/Disease_engine_project/release_evidences.json") #Path where the file is located
```

```python
evidences = evidences.T.reset_index(drop=True)
```

```python
evidences = evidences[["name", "default_value", "possible-values", "data_type"]]
```

```python
features = {}
features["Diagnosis"] = []
features["Age"] = []
for i in range(0, evidences.shape[0]):
  if not evidences.iloc[i, 2]:
    features[evidences.iloc[i, 0]] = []
    # If the feature is a binary question, it is stored as it is
  else:
    for j in evidences.iloc[i, 2]:
      features[evidences.iloc[i,0] + "_" + str(j)] = []
      ## if the feature has multiple options, store each evidence_option pair as one feature
## Features are extracted from the evidences dataset. Each evidence is a feature that will be utilized for transforming patient
## records into binary vectors
```

```python
feature_df = pd.DataFrame(features)
```
