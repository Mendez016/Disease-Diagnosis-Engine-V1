```python
import pandas as pd
```

```python
import duckdb
```

```python
conditions = pd.read_json("/content/drive/MyDrive/Disease_engine_project/release_conditions.json")
```

```python
con_names = conditions.columns
```

```python
conn = duckdb.connect()
```

```python
feature_df = pd.read_csv("/content/drive/MyDrive/Disease_engine_project/processed_data/features.csv")
```

```python
query = "SELECT * FROM '/content/drive/MyDrive/Disease_engine_project/processed_data/output_*.parquet' WHERE Diagnosis = \'"
## SQL query to pull the data from the specified disease across all the parquets
averages_df = feature_df.iloc[:,:].drop(columns=["Age"])

for i in con_names:
  ## Get averaged vectors (centroids) for each condition
  condition_df = conn.execute(query + i + "\'").df().drop(columns=["Diagnosis", "Age"])
  avg_row = {}
  avg_row["Diagnosis"] = [i]
  for j in condition_df:
    avg_row[j] = [condition_df[j].mean()] ## Get average for evidence
  avg_row = pd.DataFrame(avg_row)
  averages_df = pd.concat([averages_df, avg_row], ignore_index=True)

averages_df.to_parquet("/content/drive/MyDrive/Disease_engine_project/processed_data/output_average.parquet", engine='pyarrow')
```
